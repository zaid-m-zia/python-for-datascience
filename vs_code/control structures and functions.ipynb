```python
"""
A common thing to do with your dataset (which looks like health/fitness data) 
is to categorize Calories or
Pulse. Let's categorize Calories as "Low Burn" or " Moderate Burn ".
"""
for i in range(0,len(sample_data),1):
    if (sample_data['Calories'][i]<=350):
        sample_data['Status'][i]='Low Burn'
    elif (sample_data['Calories'][i]>=450):
        sample_data['Status'][i]='High Burn'
    else:
        sample_data['Status'][i]='Moderate Burn'
```
